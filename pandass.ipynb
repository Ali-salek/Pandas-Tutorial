# How to handl missing data in Pandas

```python
missing_values = ["n.a." , "NA" , "na" , 0] # Standardized Missing Values as Pandas Format
df = pd.read_csv("out1.csv" , na_values= missing_values)
df.head(6)
# two solutions for handle missing values : 1. Remove the rows which has/have missing values 2.filling out the missing values
```

```python
# removing solution (dropna)
df.dropna(axis = 0 , inplace= True , how="all")
```

```python
df
```

```python
df["Salary"].fillna(df["Salary"].mean() , inplace = True) #filling with median , mode , mean
df["Salary"].fillna(1111 , inplace = True) # filling with constant value
df["Salary"].fillna(method = "ffill" , inplace = True) # filling with forward value
df["Salary"].fillna(method = "bfill" , inplace = True) # filling with backward value
df["Salary"].interpolate(method = "linear") # filling with linear interpolation
```

```python
df["Salary"]=df["Salary"].interpolate(method="linear") # filling with linear interpolation
# x1 = (x0 + x2) / 2
```

# How to join Tabels in pandas

```python
# Join
# join is primarily used when you want to combine DataFrames based on their index.
df11 = pd.DataFrame({
        'A':['A0','A1','A2','A3','A4','A5'],
    })

df12 = pd.DataFrame({
        'B':['B0','B1','B2'],
    })
```

```python
df11.join(df12)
```

```python
# Merge Function
# When two DataFrames share a common column and you want to join them based on it.
dfm1 = pd.DataFrame({
        'key':['A','B','C','F'],
        'value1':[1,2,3,4],
    })

dfm2 = pd.DataFrame({
        'key':['A','B','C','D'],
        'value2':[4,5,6,7],
    })
```

```python
new_df = pd.merge(dfm1, dfm2, on="key")
new_df
```

```python
new_df= pd.merge(dfm1, dfm2, on="key", how="outer")
new_df
```

```python
new_df= pd.merge(dfm1, dfm2, on="key", how="left")
new_df
```

```python
# Concat Function 
# concat is primarily used to join DataFrames together, either vertically or horizontally.
dfc1 = pd.DataFrame({
        'key':['B','B','A','C'],
        'value':[1,2,3,4],
    })

dfc2 = pd.DataFrame({
        'key':['A','B' ,'D'],
        'value':[1,2,3],
    })

# row concatenation
pd.concat([dfc1, dfc2])
```

```python
# columns concatination
pd.concat([dfc1, dfc2], axis=1)
```
